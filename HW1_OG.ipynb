# **ЧАСТЬ 1. Работа со строками и исключениям**

```python
def parse_int(s:str) -> int | None:
  s=s.strip()
  try:
    return int(s)
  except ValueError:
      return None
```

Здесь мы задали функцию, которая принимает в себя строку и преобразует ее в целочисленное значение, а если это невозможно - выдает *`None`*. Кроме того функция убирает пробелы в начале и в конце полученной строки.

Далее сделаем проверку.


```python

print("1 случай -", parse_int("12"))
print("2 случай -", parse_int(" -7 "))
print("3 случай -", parse_int("12a"))
print("4 случай -", parse_int(""))
```

Всё гуд, целочисленные значения 12 и -7 вывелись корректно, а когда затисалась буква и просто не было чисел вывело None. Далее будет ячейка для универсальной проверки, чтобы можно было ввести любую другую строчку.

```python

print("Ваш случай -", parse_int(input("введите лбую строчку - ")))
```

**Функция 1. count_positive**

```python

def count_positive(nums: list[int]) -> int:
    count = 0

    for i in nums:
        if i > 0:
            count += 1

    return count
```

Функция считает количество чисел в списке, которые строго больше нуля, то есть считает положительные числа. С помощью оператора `for` функция бежит по списку, если видит положительное число, то прибавляет 1 к счётчику `count`. Если в списке не было бы положительных чисел, то функция бы вывела 0.

Далее проверка по основному и проверочному списку.

```python
print ("Основной список - ", count_positive(my_list_1))
print ("Проверочный список - ", count_positive(my_list_2))
```

Всё гуд, числа считаются корректно.

**Функция 2. sum_even**

```python

def sum_even(nums: list[int]) -> int:
    my_sum = 0
    for i in nums:
        if i % 2 == 0:
            my_sum += i
    return my_sum
```

Функция суммирует все чётные чила в списке. С помощью оператора `for` она бежит по списку, если видит чётное число, то прибавляет его значение к сумме `my_sum`. Если бы чётных чисел не было, то опять же был бы 0 в конце.

Далее проверка.

```python
print ("Основной список - ", sum_even(my_list_1))
print ("Проверочный список - ", sum_even(my_list_2))
```

Тоже всё гуд, сумма корректная.

**Функция 3. first_visible**

```python

def first_divisible(nums: list[int], k: int) -> int | None:
    my_number = None
    for i in nums:
        if i % k == 0:
            my_number = i
            break
    return my_number
```

Функция всё так же бежит по списку, пока не наткнется на число `i`, которое делится на некое число `k` без остатка. Как только фунция натыкается на такое число `i`, цикл прерывается с помощью `break` и дальше функция не бежит, а просто выводит число, на котором она остановилась. Если такого числа вообще нет, то выводится `None` в конце.

Далее проверка.

```python

print ("Основной список - ", first_divisible(my_list_1, int(input("Введите число k - "))))
print ("Проверочный список -", first_divisible(my_list_2, int(input("Введите число K - "))))

```

Тоже всё гуд, функция выодит корректное число или None, если такого числа вообще нет.

# **ЧАСТЬ 3. Lambda и сортировки**

Для начала заводим наш список словарей.


```python

items = [
  {"name": "A", "price": 120, "rating": 4.7},
  {"name": "B", "price": 90, "rating": 4.2},
  {"name": "C", "price": 150, "rating": 4.9},
]
```

1) Теперь сортируем список по цене. (на возрастаниен)

```python

sorted_by_price = sorted(items, key=lambda x: x["price"])
sorted_by_price

```

Отсортировали список с помощью функции `sorted()` и маленькой функции `lambda`, которая берет элемент списка (в нашем случае элементом является словарь) и возвращает его `price`. По умолчанию сортировка идет на возрастание.


2. Сортировка по рейтингу. (на убывание)

```python

sorted_by_rating = sorted(items, key=lambda x: -x["rating"])
sorted_by_rating
```

Опять отсортировали список с помощью функции `sorted()` и маленькой функции `lambda`, которая берет элемент списка  и возвращает его `rating`, НО со знаком минус, чтобы чем выше был рейтинг, тем выше он был в списке, так как по умолчанию `sorted` ставит в начале самое маленькое значение.

# Часть 4. Мини ООП

Создаём класс `Chocolate`




```python
class Chocolate:
  def __init__(self, name: str, price: float):
    if price <=0:
      raise ValueError("Слишком дешево, мы так не заработаем")
    self.name = name
    self.price = price

  def discount(self, percent: int) -> float:
    lower_price = self.price * (1 - percent / 100)
    return lower_price
```

Инициализировали обьект с двумя атрибутами: строчный name и числовой price. Также немного обезопасили себя от некорректной цены, которая не может быть отрицательной или нулевой с помощью проверки if и ValueError. Дальше сохраняем данные внутри обьекта self.name = name. Далее сразу создаем метод, который будет считать нам скидку на основе цены положенной в объект и скидки, которую мы зададим.

**Проверка класса и метода**

```python
chocolate_1 = Chocolate("ВкусВилл", 4900.0)
chocolate_2 = Chocolate("Красная Цена", 12.6)

print(chocolate_1.name, "- уцененка 1% - ", chocolate_1.discount(1))
print(chocolate_2.name, "- уцененка 95% - ", chocolate_2.discount(95))
```
